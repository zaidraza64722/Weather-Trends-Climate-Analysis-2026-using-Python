```python
# Convert the Date Column into Datetime Format

df["time"] = pd.to_datetime(df["time"])
```

```python
df.dtypes
```

```python
# Extract the Year from the Date Column

df["year"] = df["time"].dt.year
```

```python
df[["time", "year"]].head(10)
```

```python
# Calculate average, maximum and minimum temperatures for each city.

df.groupby("city")["temperature_2m_mean (°C)"].agg(
    ["mean", "max", "min", "std"]
)
```

```python
# highest mean temperature and its date for each city

max_temp = df.loc[
    df.groupby("city")["temperature_2m_mean (°C)"].idxmax(),
    ["city", "time", "temperature_2m_mean (°C)"]
]

max_temp
```

```python
min_temp = df.loc[
    df.groupby("city")["temperature_2m_mean (°C)"].idxmin(),
    ["city", "time", "temperature_2m_mean (°C)"]
]

min_temp
```

```python
# Combine both results

extreme_weather = max_temp.merge(
    min_temp,
    on="city",
    suffixes=("_highest", "_lowest")
)

extreme_weather
```

```python
# Calculate city-wise weather statistics

weather_summary = df.groupby("city").agg(
    avg_temperature=("temperature_2m_mean (°C)", "mean"),
    max_temperature=("temperature_2m_max (°C)", "max"),
    min_temperature=("temperature_2m_min (°C)", "min"),
    avg_rainfall=("precipitation_sum (mm)", "mean"),
    total_rainfall=("precipitation_sum (mm)", "sum"),
    avg_wind_speed=("wind_speed_10m_max (km/h)", "mean")
)

weather_summary
```

```python
df["month"] = df["time"].dt.month
```

```python
# Calculate monthly average temperature for each city

monthly_temperature = pd.pivot_table(
    df,
    values="temperature_2m_mean (°C)",
    index="city",
    columns="month",
    aggfunc="mean"
)

monthly_temperature
```

```python
# Calculate average temperature by city

city_ranking = (
    df.groupby("city")["temperature_2m_mean (°C)"]
    .mean()
    .reset_index()
)
```

```python
# Calculate average temperature by city and apply rank

city_ranking["temperature_rank"] = (
    city_ranking["temperature_2m_mean (°C)"]
    .rank(ascending=False, method="dense")
)
city_ranking
```

```python
# Sort the ranking

sort= city_ranking.sort_values(
    by="temperature_rank"
)

sort
```
