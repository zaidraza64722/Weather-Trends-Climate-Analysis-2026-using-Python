```python
import pandas as pd

df = pd.read_excel("../Dataset/Cleaned/weather_2026_5_cities_cleaned.xlsx")

df["time"] = pd.to_datetime(df["time"])
df["month"] = df["time"].dt.month

df.head()
```

```python
# Average Temperature Comparison Across Cities

city_summary = df.groupby("city")["temperature_2m_mean (°C)"].mean()

print("Average Temperature by City:")
print(city_summary)

print("\nHottest City:")
print(city_summary.idxmax())

print("\nCoolest City:")
print(city_summary.idxmin())
```

```python
# Total Rainfall Comparison Across Cities

rainfall_summary = df.groupby("city")["precipitation_sum (mm)"].sum()

print("Total Rainfall by City:")
print(rainfall_summary)

print("\nHighest Rainfall City:")
print(rainfall_summary.idxmax())

print("\nLowest Rainfall City:")
print(rainfall_summary.idxmin())
```

```python
# Monthly Temperature Trends Across Cities.
# Hottest and Coolest Month for Each City

monthly_summary = df.groupby(
    ["city", "month"]
)["temperature_2m_mean (°C)"].mean().reset_index()

hottest_month = monthly_summary.loc[
    monthly_summary.groupby("city")["temperature_2m_mean (°C)"].idxmax()
]

coolest_month = monthly_summary.loc[
    monthly_summary.groupby("city")["temperature_2m_mean (°C)"].idxmin()
]

print("Hottest Month:")
print(hottest_month)

print("\nCoolest Month:")
print(coolest_month)
```

```python
# Average Wind Speed Comparison Across Cities

wind_summary = df.groupby("city")["wind_speed_10m_max (km/h)"].mean()

print("Average Wind Speed by City:")
print(wind_summary)

print("\nHighest Wind Speed City:")
print(wind_summary.idxmax())

print("\nLowest Wind Speed City:")
print(wind_summary.idxmin())
```

```python
# Highest and Lowest Temperature Recorded

highest = df.loc[df["temperature_2m_mean (°C)"].idxmax()]

lowest = df.loc[df["temperature_2m_mean (°C)"].idxmin()]

print("Highest Temperature:")
print(highest[["city", "time", "temperature_2m_mean (°C)"]])

print("\nLowest Temperature:")
print(lowest[["city", "time", "temperature_2m_mean (°C)"]])
```

```python
# Number of Rainy Days Across Cities

rainy_days = df[df["precipitation_sum (mm)"] > 0]

rainy_summary = rainy_days.groupby("city")["time"].count()

print("Rainy Days by City:")
print(rainy_summary)

print("\nCity with Most Rainy Days:")
print(rainy_summary.idxmax())

print("\nCity with Fewest Rainy Days:")
print(rainy_summary.idxmin())
```

```python
# Temperature Range Comparison Across Cities

temperature_range = df.groupby("city")["temperature_2m_mean (°C)"].agg(
    ["max", "min"]
)

temperature_range["range"] = (
    temperature_range["max"] - temperature_range["min"]
)

print("Temperature Range by City:")
print(temperature_range)

print("\nCity with Highest Temperature Variation:")
print(temperature_range["range"].idxmax())

print("\nCity with Lowest Temperature Variation:")
print(temperature_range["range"].idxmin())
```

```python
# Average Wind Gust Comparison Across Cities

gust_summary = df.groupby("city")["wind_gusts_10m_max (km/h)"].mean()

print("Average Wind Gust by City:")
print(gust_summary)

print("\nCity with Highest Average Wind Gust:")
print(gust_summary.idxmax())

print("\nCity with Lowest Average Wind Gust:")
print(gust_summary.idxmin())
```

```python
# Highest and Lowest Rainfall Month for Each City

monthly_rainfall = df.groupby(
    ["city", "month"]
)["precipitation_sum (mm)"].sum().reset_index()

highest_rainfall = monthly_rainfall.loc[
    monthly_rainfall.groupby("city")["precipitation_sum (mm)"].idxmax()
]

lowest_rainfall = monthly_rainfall.loc[
    monthly_rainfall.groupby("city")["precipitation_sum (mm)"].idxmin()
]

print("Month with Highest Rainfall:")
print(highest_rainfall)

print("\nMonth with Lowest Rainfall:")
print(lowest_rainfall)
```
