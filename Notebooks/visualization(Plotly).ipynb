```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_excel("../Dataset/Cleaned/weather_2026_5_cities_cleaned.xlsx")

df["time"] = pd.to_datetime(df["time"])

df["month"] = df["time"].dt.month
```

```python
# How do monthly average temperatures compare across all five cities?

fig = px.line(
    df,
    x="time",
    y="temperature_2m_mean (°C)",
    color="city",
    title="Daily Temperature Trends – Five Cities"
)

fig.show()
```

```python
# How does average temperature differ geographically across our five Indian cities?

# 1. Add city coordinates
city_coordinates = {
    "Bengaluru": [12.9719, 77.5937],
    "Mumbai": [19.0728, 72.8826],
    "Delhi": [28.6519, 77.2315],
    "Chennai": [13.0878, 80.2785],
    "Hyderabad": [17.3840, 78.4564]
}

df["latitude"] = df["city"].map(lambda x: city_coordinates[x][0])
df["longitude"] = df["city"].map(lambda x: city_coordinates[x][1])

# 2. Calculate average temperature by city
city_temp = df.groupby("city").agg(
    avg_temperature=("temperature_2m_mean (°C)", "mean"),
    latitude=("latitude", "first"),
    longitude=("longitude", "first")
).reset_index()

# 3. Create the interactive map

import plotly.express as px

fig = px.scatter_mapbox(
    city_temp,
    lat="latitude",
    lon="longitude",
    color="avg_temperature",
    size="avg_temperature",
    hover_name="city",
    zoom=4,
    title="Average Temperature Across Five Indian Cities",
    mapbox_style="open-street-map"
)

fig.show()
```

```python
# compares the average temperature of each city month by month.

monthly_temp = df.groupby(
    ["city", "month"]
)["temperature_2m_mean (°C)"].mean().reset_index()

fig = px.bar(
    monthly_temp,
    x="month",
    y="temperature_2m_mean (°C)",
    color="city",
    barmode="group",
    title="Monthly Average Temperature by City"
)

fig.show()
```

```python
# shows how frequently different temperature ranges occur across all five cities.

fig = px.histogram(
    df,
    x="temperature_2m_mean (°C)",
    color="city",
    nbins=20,
    title="Temperature Distribution Across Five Cities"
)

fig.show()
```

```python
# compares the temperature distribution across all five cities and helps identify unusually high or low temperatures.

fig = px.box(
    df,
    x="city",
    y="temperature_2m_mean (°C)",
    color="city",
    title="Temperature Distribution by City"
)

fig.show()
```
